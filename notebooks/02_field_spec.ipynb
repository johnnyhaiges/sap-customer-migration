## Record-Level Validation Rules
R1: Contactability: At least one of [phone, email] must be non-null after cleansing. If both are null, REJECT RECORD (R).
R2: [account_group] is mandatory due to its role in SAP's customer master data framework. Any accounts without it must be rejected. 
R3: For [postal_code] FIX RECORD (F) values based off of country's respective postal format, if a row does not have a country, REJECT RECORD (R).
R4: For all columns labeled with only FIX RECORD (F) if there is no value present default, list is as follows.
[currency]: derive by country 
[language]: derive by country 
[sales_org]: default to most common value (1000) if unable to confirm with business
[payment_terms]: default to NT00, most conervative credit terms, a la, payment due immediately.
[credit_limit]: default to 0, conservative estimate.
[division]: default to 00, most common value.
[created_date]: parse default, anything that can be parsed is rejected.
