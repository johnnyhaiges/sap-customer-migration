```python
def parse_date(val):
    if pd.isna(val):
        return pd.NaT
    val = str(val).strip()
    for fmt in ['%Y-%m-%d', '%Y/%m/%d', '%d/%m/%Y', '%m-%d-%Y', '%d.%m.%Y', '%b %d, %Y']:
        try:
            return pd.to_datetime(val, format=fmt)
        except (ValueError, TypeError):
            continue
    return pd.NaT  # unparseable → null

cust['created_date'] = cust['created_date'].apply(parse_date).dt.strftime('%Y-%m-%d')
```

```python
print('Null count after parsing:', cust['created_date'].isnull().sum())
print('Sample:')
print(cust['created_date'].dropna().sample(15, random_state=4).tolist())
# Check format consistency: all should be 10 chars (YYYY-MM-DD)
print('Length check:', cust['created_date'].dropna().str.len().value_counts())
```

```python
name_norm = (cust['first_name'].str.lower().str.strip().fillna('') + '|' +
             cust['last_name'].str.lower().str.strip().fillna(''))
```

```python
identifier = cust['email'].fillna(cust['tax_id']).fillna(cust['phone'])
```

```python
fingerprint = name_norm + '|' + identifier
```

```python
is_dup = fingerprint.duplicated(keep='first')

print('Records flagged as duplicates:', is_dup.sum())
print('Records with no usable fingerprint (all 3 identifiers null):',
      identifier.isnull().sum())
```

```python
dup_fingerprints = fingerprint[is_dup].value_counts().head(10)
print('Most-duplicated fingerprints:')
print(dup_fingerprints)
```

```python
n_clusters = (fingerprint.value_counts() > 1).sum()
print('Distinct duplicate clusters:', n_clusters)
print('Total records in duplicate clusters:', is_dup.sum() + n_clusters)
print('Records to be kept (one per cluster):', n_clusters)
print('Records to be dropped:', is_dup.sum())
```

```python
cust['_duplicate'] = is_dup
```

```python
cust = cust[~cust['_duplicate']].copy()
print('Records remaining:', len(cust))
```

```python
reasons = [[] for _ in range(len(cust))]
cust = cust.reset_index(drop=True)
```

```python
required_fields = {
    'customer_id': 'missing customer_id',
    'first_name':  'missing first_name',
    'last_name':   'missing last_name',
    'city':        'missing city',
    'street':      'missing street',
}
for field, msg in required_fields.items():
    bad = cust[field].isnull()
    for i in cust.index[bad]:
        reasons[i].append(msg)
```

```python
no_contact = cust['phone'].isnull() & cust['email'].isnull()
for i in cust.index[no_contact]:
    reasons[i].append('R1: no phone or email')
```

```python
no_acct = cust['account_group'].isnull()
for i in cust.index[no_acct]:
    reasons[i].append('R2: missing account_group')
```

```python
no_country = cust['country'].isnull()
for i in cust.index[no_country]:
    reasons[i].append('R3: missing country')
```

```python
cust['reject_reason'] = ['; '.join(r) if r else '' for r in reasons]
```

```python
rejected = cust[cust['reject_reason'] != ''].copy()
clean    = cust[cust['reject_reason'] == ''].copy()
```

```python
print(f'Clean records: {len(clean):,}')
print(f'Rejected records: {len(rejected):,}')
print(f'Total: {len(clean) + len(rejected):,}')
print()
print('Top rejection reasons:')
print(rejected['reject_reason'].value_counts().head(10))
```
